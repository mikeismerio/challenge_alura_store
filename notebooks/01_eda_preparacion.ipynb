```python
# =====================================
# 🧠 Challenge Alura Store - Análisis de Tiendas
# =====================================

import pandas as pd
from pathlib import Path
import janitor

# -------------------------------------
# Definir ruta de los datos RAW
# -------------------------------------
raw_path = Path("../data/raw")

# -------------------------------------
# Carga de los archivos CSV por tienda
# -------------------------------------
tienda1 = pd.read_csv(raw_path / "tienda_1.csv")
tienda2 = pd.read_csv(raw_path / "tienda_2.csv")
tienda3 = pd.read_csv(raw_path / "tienda_3.csv")
tienda4 = pd.read_csv(raw_path / "tienda_4.csv")
```

```python
# Vista rápida
tienda1.head()
```

```python
# Limpia nombres de columnas automáticamente
tienda1 = tienda1.clean_names()
tienda2 = tienda2.clean_names()
tienda3 = tienda3.clean_names()
tienda4 = tienda4.clean_names()
```

```python
# Verificar nombres limpios
tienda1.columns
```

```python
# -------------------------------------
# Ejemplo de inspección: valores únicos de método_de_pago
# -------------------------------------
tienda1['metodo_de_pago'].unique()
```

```python
# -------------------------------------
# Comparar estructura de columnas entre archivos
# -------------------------------------
tabla_columnas = pd.DataFrame({
"Tienda 1": pd.Series(tienda1.columns),
"Tienda 2": pd.Series(tienda2.columns),
"Tienda 3": pd.Series(tienda3.columns),
"Tienda 4": pd.Series(tienda4.columns),
})
# Mostrar la tabla comparativa
tabla_columnas
```

```python
# -------------------------------------
# Validar dimensiones de cada dataset
# -------------------------------------
print(tienda1.shape)
print(tienda2.shape)
print(tienda3.shape)
print(tienda4.shape)
```

```python
# -------------------------------------
# Número de lugares distintos de compra por tienda
# -------------------------------------
print(tienda1['lugar_de_compra'].nunique())
print(tienda2['lugar_de_compra'].nunique())
print(tienda3['lugar_de_compra'].nunique())
print(tienda4['lugar_de_compra'].nunique())
```

```python
# -------------------------------------
# Estadísticas descriptivas generales
# -------------------------------------
tienda1.describe(include='all')
```

```python
# -------------------------------------
# Conteo de valores nulos por columna
# -------------------------------------
tienda1.isnull().sum()
```

```python
# -------------------------------------
# Unificación de todas las tiendas
# -------------------------------------
tienda1["tienda"] = "Tienda 1"
tienda2["tienda"] = "Tienda 2"
tienda3["tienda"] = "Tienda 3"
tienda4["tienda"] = "Tienda 4"

df = pd.concat([tienda1, tienda2, tienda3, tienda4], ignore_index=True)
```

```python
# -------------------------------------
# Guardar dataset como CSV procesado
# -------------------------------------
processed_path = Path("../data/processed")
processed_path.mkdir(parents=True, exist_ok=True)

df.to_csv(processed_path / "tiendas.csv", index=False, encoding="utf-8")

print("✅ Archivo guardado en data/processed/tiendas.csv")
```
